# 3. Static types and variables

C++ is statically typed. This means a variable must be declared with a type before it is used.

| Code | Meaning |
|---|---|
| `int x;` | create an integer variable named `x` |
| `x = 42;` | store `42` inside `x` |
| `int x = 42;` | create `x` and store `42` immediately |
| `double v = 3.3;` | create a decimal number variable |
| `bool alarm = false;` | create a true/false variable |
| `char grade = 'A';` | create a single-character variable |

A useful beginner habit: choose names that explain what the value means, such as `voltage`, `count`, or `temperature_c`.

# 5. Arrays

An array stores several values of the same type in a fixed-size block.

Important rules:

1. Array indices start at `0`.
2. A length-3 array has positions `0`, `1`, and `2`.
3. Raw C++ arrays do not print nicely by just writing the array name.
4. `sizeof(array)` returns the total number of bytes used by the whole array.

| Code | Meaning |
|---|---|
| `int scores[3] = {10, 20, 30};` | create a 3-element integer array |
| `scores[0]` | first value |
| `scores[2]` | third value |
| `scores[1] = 99;` | replace the second value |


## Example 5.1 — Create and read an array


````c++
int scores[3] = {10, 20, 30};
scores[0]
````

````c++
scores[2]
````

# 7. Structs: grouping related values

A `struct` lets you create a new type by grouping related fields.

This is better than a plain array when the values have different meanings.

For a birthday, this:

```cpp
Birthday bday = {1992, 5, 8};
```

is clearer than this:

```cpp
short birthday_array[3] = {1992, 5, 8};
```

because `bday.year`, `bday.month`, and `bday.day` explain the meaning of each value.


## Example 7.1 — Define a `Birthday` struct


````c++
struct Birthday {
    short year;
    unsigned char month;
    unsigned char day;
};
````

## Exercise 7.1 — Store one birthday


````c++
Birthday bday1 = {1992, 5, 8};
bday1.year
````

## Exercise 7.2 — Access the month field


````c++
+bday1.month
````

## Exercise 7.3 — Access the day field


````c++
+bday1.day
````

## Exercise 7.4 — Size of a struct

Find how many bytes `bday1` uses. The result may be larger than the sum of the fields because C++ may add padding for efficient memory access.


````c++
sizeof(bday1)
````

## Exercise 7.5 — Array of structs

Create an array called `bdays` containing three birthdays.


````c++
Birthday bdays[3] = {
    {1992, 5, 8},
    {2000, 10, 1},
    {2004, 12, 25}
};
+bdays[0].month
````

## Exercise 7.6 — Print the day of the second birthday


````c++
+bdays[1].day
````

## Exercise 7.7 — Print the year of the third birthday


````c++
bdays[2].year
````

## Exercise 7.8 — Size of the birthday array


````c++
sizeof(bdays)
````
