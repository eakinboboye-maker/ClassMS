## 1. Functions: naming a reusable piece of work

A function is a named block of code.

Think of a function as a small machine:

```text
inputs  --->  function  --->  output
```

A C/C++ function normally has this shape:

```cpp
return_type function_name(parameter_type parameter_name) {
    statements;
    return value;
}
```

Example:

```cpp

```

means:

- the function is called `add2`;
- it receives two `int` inputs called `x` and `y`;
- it returns one `int` result.

````c++
int add2(int x, int y) {
    int z = x + y;
    return z;
}

int add_example_result = add2(3, 5);
add_example_result
````

## 2. Scope and side effects

A variable declared inside a function belongs to that function. This is called a **local variable**.

A variable declared outside all functions is called a **global variable**. Functions can access global variables, but you should avoid using many global variables in serious programs because they make bugs harder to find.

For now, we use a global variable to understand how a function can change program state.

````c++
int counter = 0;
````

### Important idea: return value vs printing

These two functions are different:

```cpp
int square(int x)       // calculates and returns a value
void printSquare(int x) // prints something, but returns nothing
```

In beginner programs, prefer returning values when possible. Returned values are easier to test.

````c++
int square(int x) {
    return x * x;
}

void printSquare(int x) {
    printf("%d\n", x * x);
}

square(6) == 36
````

## 4. `while` loops

A `while` loop repeats while a condition remains true.

```cpp
while (condition) {
    // repeated code
}
```

A loop usually needs three parts:

1. **Start**: create a variable before the loop.
2. **Test**: check whether the loop should continue.
3. **Update**: change the variable inside the loop.

If you forget the update, you may create an infinite loop.

````c++
int while_demo_n = 0;

while (while_demo_n < 5) {
    while_demo_n = while_demo_n + 1;
    printf("%d\n", while_demo_n);
}
````

### Example 3: running total with `while`

The running total of `5` is:

```text
1 + 2 + 3 + 4 + 5 = 15
```

````c++
int runningTotal(int n) {
    int value = 1;
    int total = 0;

    while (value <= n) {
        total = total + value;
        value = value + 1;
    }

    return total;
}
````

````c++
runningTotal(5) == 15
````

````c++
runningTotal(10) == 55
````

### Exercise 7: sum an array with `while`

Write a function called `arraySum`.

Requirements:

- inputs: an integer array and its size;
- output: sum of all array elements;
- use a `while` loop.

````c++
int arraySum(const int array[], int size) {
    int index = 0;
    int total = 0;

    while (index < size) {
        total = total + array[index];
        index = index + 1;
    }

    return total;
}
````

````c++
int arraySum_test_values1[3] = {1, 2, 3};
arraySum(arraySum_test_values1, 3) == 6
````

````c++
int arraySum_test_values2[5] = {5, 10, 15, 20, 25};
arraySum(arraySum_test_values2, 5) == 75
````

### Exercise 8: find the maximum value in an array

Write a function called `arrayMax`.

Requirements:

- inputs: an integer array and its size;
- output: the largest value;
- assume `size` is at least `1`.

````c++
int arrayMax(const int array[], int size) {
    int maximum = array[0];
    int index = 1;

    while (index < size) {
        if (array[index] > maximum) {
            maximum = array[index];
        }
        index = index + 1;
    }

    return maximum;
}
````

````c++
int arrayMax_test_values1[5] = {3, 9, 2, 11, 4};
arrayMax(arrayMax_test_values1, 5) == 11
````

````c++
int arrayMax_test_values2[4] = {-8, -2, -10, -5};
arrayMax(arrayMax_test_values2, 4) == -2
````

### Exercise 11: count even numbers

Write a function called `countEven`.

Requirements:

- inputs: an integer array and its size;
- output: number of even values in the array;
- use `%` modulo operator.

Reminder:

```cpp
x % 2 == 0
```

means `x` is divisible by `2`.

````c++
int countEven(const int array[], int size) {
    int count = 0;

    for (int index = 0; index < size; index = index + 1) {
        if (array[index] % 2 == 0) {
            count = count + 1;
        }
    }

    return count;
}
````

### Exercise 12: average sensor reading

Write `averageReading`.

Requirements:

- input: integer array and its size;
- output: average as a `double`;
- if `size <= 0`, return `0.0`;
- use your loop knowledge.

````c++
double averageReading(const int readings[], int size) {
    if (size <= 0) {
        return 0.0;
    }

    int total = 0;

    for (int index = 0; index < size; index = index + 1) {
        total = total + readings[index];
    }

    return (double) total / size;
}
````

### Exercise 13: count readings above a limit

Write `countAboveLimit`.

Requirements:

- inputs: integer array, size, and integer limit;
- output: how many readings are greater than the limit.

````c++
int countAboveLimit(const int readings[], int size, int limit) {
    int count = 0;

    for (int index = 0; index < size; index = index + 1) {
        if (readings[index] > limit) {
            count = count + 1;
        }
    }

    return count;
}
````

## 9. Optional challenge

Write a function called `isSortedAscending`.

Requirements:

- input: integer array and its size;
- return `true` if each value is less than or equal to the value after it;
- return `false` otherwise.

Examples:

```cpp
{1, 2, 2, 4} -> true
{1, 5, 3, 7} -> false
```

````c++
bool isSortedAscending(const int array[], int size) {
    for (int index = 0; index < size - 1; index = index + 1) {
        if (array[index] > array[index + 1]) {
            return false;
        }
    }

    return true;
}
````

````c++
int sorted_test_values1[4] = {1, 2, 2, 4};
isSortedAscending(sorted_test_values1, 4) == true
````

````c++
int sorted_test_values2[4] = {1, 5, 3, 7};
isSortedAscending(sorted_test_values2, 4) == false
````
