## 1. Source code, compiler, executable

A C++ program starts as **source code**, which is a human-readable text file such as `hello.cpp`.

The computer does not directly run this source code. A compiler translates it into an **executable** file.

Example:

```text
hello.cpp  --compiled by g++-->  hello.out  --run by OS-->  output on screen
```

### The smallest useful C++ program

```cpp
#include <cstdio>

int main() {
    printf("Hello, World!\n");
    return 0;
}
```

| Part | Meaning |
|---|---|
| `#include <cstdio>` | Gives access to `printf()` |
| `int main()` | The starting point of the program |
| `{ ... }` | The body of the function |
| `printf("...\n")` | Prints text to the terminal |
| `return 0;` | Tells the operating system that the program ended successfully |


````c++
!cat > hello.cpp <<'CPP'
#include <cstdio>

int main() {
    printf("Hello, World!\n");
    return 0;
}
CPP
````

````c++
!g++ -std=c++17 -Wall -Wextra hello.cpp -o hello.out && ./hello.out
````

## 2. Printing values with `printf`

`printf()` can print text and values. It uses **format specifiers** to know the type of value to print.

| Format | Type | Example |
|---|---|---|
| `%d` | integer | `printf("%d\n", 25);` |
| `%f` | floating-point number | `printf("%f\n", 3.14);` |
| `%.2f` | floating-point number with 2 decimal places | `printf("%.2f\n", 3.14);` |
| `%c` | character | `printf("%c\n", 'A');` |
| `%s` | C-style string | `printf("%s\n", "Hello");` |

### Engineering example: electrical power

Power is:

```text
P = V × I
```

where:

- `P` is power in watts
- `V` is voltage in volts
- `I` is current in amperes


````c++
!cat > power.cpp <<'CPP'
#include <cstdio>

int main() {
    double voltage = 12.0;
    double current = 1.5;
    double power = voltage * current;

    printf("Voltage: %.2f V\n", voltage);
    printf("Current: %.2f A\n", current);
    printf("Power: %.2f W\n", power);

    return 0;
}
CPP
````

````c++
!g++ -std=c++17 -Wall -Wextra power.cpp -o power.out && ./power.out
````

## 3. Writing functions

A function is a named block of code that performs a specific job.

General form:

```cpp
return_type function_name(parameter_type parameter_name) {
    // work
    return result;
}
```

Example:

```cpp
int multiply(int x, int y) {
    return x * y;
}
```

### Why functions matter

Functions help us:

- avoid repeating code,
- test one idea at a time,
- give meaningful names to calculations,
- build large programs from small reliable parts.

### Function vocabulary

| Word | Meaning |
|---|---|
| Function definition | The actual code of the function |
| Function call | Using the function, e.g. `multiply(3, 4)` |
| Parameter | A variable listed in the function definition |
| Argument | The actual value passed during a function call |
| Return value | The result sent back by the function |


````c++
!cat > functions_example.cpp <<'CPP'
#include <cstdio>

int multiply(int x, int y) {
    return x * y;
}

double electrical_power(double voltage, double current) {
    return voltage * current;
}

int main() {
    int result = multiply(3, 4);
    double power = electrical_power(12.0, 1.5);

    printf("3 multiplied by 4 is %d\n", result);
    printf("Power is %.2f W\n", power);

    return 0;
}
CPP
````

````c++
!g++ -std=c++17 -Wall -Wextra functions_example.cpp -o functions_example.out && ./functions_example.out
````

## 4. Header files: separating interface from implementation

As programs grow, we often separate code into multiple files.

A common C/C++ structure is:

```text
math_utils.h      → declarations: what functions exist?
math_utils.cpp    → definitions: how do the functions work?
main.cpp          → program that uses the functions
```

### Important rule

A header file usually contains **declarations**, not the full program.

Good header file:

```cpp
#ifndef MATH_UTILS_H
#define MATH_UTILS_H

int square(int x);
int cube(int x);

#endif
```

The lines beginning with `#ifndef`, `#define`, and `#endif` are called an **include guard**.  
They prevent the same header file from being included more than once in a way that confuses the compiler.


````c++
!cat > math_utils.h <<'CPP'
#ifndef MATH_UTILS_H
#define MATH_UTILS_H

int square(int x);
int cube(int x);

#endif
CPP

!cat > math_utils.cpp <<'CPP'
#include "math_utils.h"

int square(int x) {
    return x * x;
}

int cube(int x) {
    return x * x * x;
}
CPP

!cat > use_math.cpp <<'CPP'
#include <cstdio>
#include "math_utils.h"

int main() {
    printf("square(5) = %d\n", square(5));
    printf("cube(3) = %d\n", cube(3));
    return 0;
}
CPP
````

````c++
!g++ -std=c++17 -Wall -Wextra use_math.cpp math_utils.cpp -o use_math.out && ./use_math.out
````

## 5. Arduino-style repeated behaviour with loops

Many embedded systems run the same task repeatedly:

```text
read sensor → decide → control output → wait → repeat
```

In Arduino, this repeated part is usually placed inside `loop()`.

In ordinary C++, we can simulate this with:

```cpp
while (true) {
    // repeated task
}
```

However, in a classroom or notebook, an uncontrolled infinite loop can freeze the session.  
So we first practise with a **limited loop**.

### Delay function

To pause a C++ program for a short time, we can use:

```cpp
#include <thread>
#include <chrono>

std::this_thread::sleep_for(std::chrono::milliseconds(500));
```


````c++
!cat > blink_limited.cpp <<'CPP'
#include <cstdio>
#include <thread>
#include <chrono>

void delay_ms(int milliseconds) {
    std::this_thread::sleep_for(std::chrono::milliseconds(milliseconds));
}

void print_led_state(bool led_on) {
    if (led_on) {
        printf("LED ON\n");
    } else {
        printf("LED OFF\n");
    }
    fflush(stdout);
}

int main() {
    for (int count = 0; count < 3; count++) {
        print_led_state(true);
        delay_ms(500);

        print_led_state(false);
        delay_ms(500);
    }

    printf("Finished limited blink simulation.\n");
    return 0;
}
CPP
````

````c++
!g++ -std=c++17 -Wall -Wextra blink_limited.cpp -o blink_limited.out && ./blink_limited.out
````

````c++
!cat > sleep_wake.cpp <<'CPP'
#include <cstdio>
#include <thread>
#include <chrono>

void delay_ms(int milliseconds) {
    std::this_thread::sleep_for(std::chrono::milliseconds(milliseconds));
}

void print_message(const char* message) {
    printf("%s\n", message);
    fflush(stdout);
}

int main() {
    for (int i = 0; i < 3; i++) {
        print_message("Sleeping...");
        delay_ms(1000);

        print_message("Awake!");
        delay_ms(500);
    }

    return 0;
}
CPP
````

## 6. Command-line arguments: giving input to a program

So far, values have been written inside the program.

But we can also provide values when running the program:

```bash
./multiply_args.out 5 3
```

Here:

- `./multiply_args.out` is the program name
- `5` is the first input
- `3` is the second input

C++ receives these through:

```cpp
int main(int argc, char* argv[])
```

| Name | Meaning |
|---|---|
| `argc` | Argument count: number of command-line parts |
| `argv` | Argument values: the command-line parts as text |
| `argv[0]` | The program name |
| `argv[1]` | First user-supplied value |
| `argv[2]` | Second user-supplied value |

Since `argv` values are text, we convert them to numbers using functions such as:

```cpp
std::atoi(argv[1]);   // text to int
std::atof(argv[1]);   // text to double
```

For beginner work, `atoi()` and `atof()` are acceptable, but later you should learn safer conversion methods.


````c++
!cat > show_args.cpp <<'CPP'
#include <cstdio>

int main(int argc, char* argv[]) {
    printf("argc = %d\n", argc);

    for (int i = 0; i < argc; i++) {
        printf("argv[%d] = %s\n", i, argv[i]);
    }

    return 0;
}
CPP
````

````c++
!g++ -std=c++17 -Wall -Wextra show_args.cpp -o show_args.out && ./show_args.out 12 7 hello
````

````c++
!cat > multiply_args.cpp <<'CPP'
#include <cstdio>
#include <cstdlib>

int multiply(int x, int y) {
    return x * y;
}

int main(int argc, char* argv[]) {
    if (argc != 3) {
        printf("Usage: ./multiply_args.out number1 number2\n");
        return 1;
    }

    int first = std::atoi(argv[1]);
    int second = std::atoi(argv[2]);

    printf("%d\n", multiply(first, second));
    return 0;
}
CPP
````

````c++
!g++ -std=c++17 -Wall -Wextra multiply_args.cpp -o multiply_args.out && ./multiply_args.out 5 3
````

## 7. Final mini-lab: simple sensor alarm

Create `sensor_alarm.cpp`.

The program should receive:

```bash
./sensor_alarm.out threshold reading1 reading2 reading3
```

Example:

```bash
./sensor_alarm.out 50 45 60 55
```

The program should:

1. Check that exactly four values were provided after the program name.
2. Convert the threshold and three readings to `double`.
3. Calculate the average of the three readings.
4. Print the threshold and average.
5. Print `ALARM` if the average is greater than or equal to the threshold.
6. Print `NORMAL` otherwise.

Recommended functions:

```cpp
double average3(double a, double b, double c);
bool alarm_triggered(double average, double threshold);
```

This mini-lab combines:

- functions,
- terminal input,
- `if/else`,
- numeric conversion,
- engineering-style decision making.


````c++
!cat > sensor_alarm.cpp <<'CPP'
#include <cstdio>
#include <cstdlib>

double average3(double a, double b, double c) {
    return (a + b + c) / 3.0;
}

bool alarm_triggered(double average, double threshold) {
    return average >= threshold;
}

int main(int argc, char* argv[]) {
    if (argc != 5) {
        printf("Usage: ./sensor_alarm.out threshold reading1 reading2 reading3\n");
        return 1;
    }

    double threshold = std::atof(argv[1]);
    double reading1 = std::atof(argv[2]);
    double reading2 = std::atof(argv[3]);
    double reading3 = std::atof(argv[4]);

    double avg = average3(reading1, reading2, reading3);

    printf("Threshold: %.2f\n", threshold);
    printf("Average: %.2f\n", avg);

    if (alarm_triggered(avg, threshold)) {
        printf("ALARM\n");
    } else {
        printf("NORMAL\n");
    }

    return 0;
}
CPP
````

````c++
!g++ -std=c++17 -Wall -Wextra sensor_alarm.cpp -o sensor_alarm.out && ./sensor_alarm.out 50 45 60 55
````

## 8. Common beginner errors checklist

Before asking for help, check these:

| Symptom | Likely cause |
|---|---|
| `expected ';'` | Missing semicolon |
| `was not declared in this scope` | Misspelled variable/function or declared after use |
| `undefined reference to ...` | Function was declared but not compiled/defined |
| Program prints nothing | Forgot to call the function or forgot `printf()` |
| `argv[1]` crashes or behaves strangely | You did not check `argc` first |
| Header included but function not found | You compiled only `main.cpp`, not the extra `.cpp` file |
| Endless loop will not stop | Use `Ctrl + C` or run with `timeout` |
