# 3. Static types and variables

C++ is statically typed. This means a variable must be declared with a type before it is used.

| Code | Meaning |
|---|---|
| `int x;` | create an integer variable named `x` |
| `x = 42;` | store `42` inside `x` |
| `int x = 42;` | create `x` and store `42` immediately |
| `double v = 3.3;` | create a decimal number variable |
| `bool alarm = false;` | create a true/false variable |
| `char grade = 'A';` | create a single-character variable |

A useful beginner habit: choose names that explain what the value means, such as `voltage`, `count`, or `temperature_c`.

# 5. Arrays

An array stores several values of the same type in a fixed-size block.

Important rules:

1. Array indices start at `0`.
2. A length-3 array has positions `0`, `1`, and `2`.
3. Raw C++ arrays do not print nicely by just writing the array name.
4. `sizeof(array)` returns the total number of bytes used by the whole array.

| Code | Meaning |
|---|---|
| `int scores[3] = {10, 20, 30};` | create a 3-element integer array |
| `scores[0]` | first value |
| `scores[2]` | third value |
| `scores[1] = 99;` | replace the second value |

# 7. Structs: grouping related values

```cpp

```

is clearer than this:

